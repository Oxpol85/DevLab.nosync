Algo que diferencia en cierto modo las funciones en el mundo de la programación, es que no sólo realizan una operación con sus entradas, sino que también parten de los siguientes principios:

- El principio de _reusabilidad_, que nos dice que si por ejemplo tenemos un fragmento de código usado en muchos sitios, la mejor solución sería pasarlo a una función. Esto nos evitaría tener código repetido, y que modificarlo fuera más fácil, ya que bastaría con cambiar la función una vez.
- Y el principio de _modularidad_, que defiende que en vez de escribir largos trozos de código, es mejor crear módulos o funciones que agrupen ciertos fragmentos de código en funcionalidades específicas, haciendo que el código resultante sea más fácil de leer.

> # ⚒ Variable de retorno "return"

El uso de la sentencia return permite realizar dos cosas:

- Salir de la función y transferir la ejecución de vuelta a donde se realizó la llamada.
- Devolver uno o varios parámetros, fruto de la ejecución de la función.

En lo relativo a lo primero, una vez se llama a return se para la ejecución de la función y se vuelve o retorna al punto donde fue llamada. Es por ello por lo que el código que va después del return no es ejecutado en el siguiente ejemplo.

`def mi_funcion():`
`print("Entra en mi_funcion")`
`return`
`print("No llega")`
`mi_funcion()`
`Output: Entra en mi_funcion`

Por ello, sólo llamamos a return una vez hemos acabado de hacer lo que teníamos que hacer en la función.

Por otro lado, se pueden devolver parámetros. Normalmente las funciones son llamadas para realizar unos cálculos en base a una entrada, por lo que es interesante poder devolver ese resultado a quien llamó a la función.

`def di_hola():`
`return "Hola"`
`di_hola()`
`Output:'Hola'`

ambién es posible devolver mas de una variable, separadas por ,. En el siguiente ejemplo tenemos una función que calcula la suma y media de tres números, y devuelve su resultado.

`def suma_y_media(a, b, c):`
`suma = a+b+c`
`media = suma/3`
`return suma, media`
`suma, media = suma_y_media(9, 6, 3)`
`print(suma)  # 18`
`print(media) # 6.0`

**Anotaciones en funciones**

Existe una funcionalidad relativamente reciente en Python llamada function annotation o anotaciones en funciones. Dicha funcionalidad nos permite añadir metadatos a las funciones, indicando los tipos esperados tanto de entrada como de salida.

Type Hinting o sugerencias de tipo) son una característica de Python que permite especificar qué tipo de datos espera recibir una función y qué tipo de dato va a devolver.

Es importante aclarar que Python sigue siendo un lenguaje de tipado dinámico. Esto significa que las anotaciones no obligan al código a cumplir con el tipo de dato (no lanzarán un error si mandas un texto en lugar de un número), pero sirven como una documentación técnica poderosa.

Sintaxis Básica
Se utilizan los dos puntos (:) para los argumentos y una flecha (->) para el resultado final:

`def calcular_impuesto(monto: float, tasa: float) -> float:`
`return monto * tasa`

- monto: float -> Indica que el argumento debe ser un número decimal.

- -> float -> Indica que la función devolverá un número decimal.

```python
import os
from datetime import datetime
from reportlab.pdfgen import canvas
from reportlab.lib.utils import ImageReader
from reportlab.lib.pagesizes import letter as A4


def generar_presupuesto(proyecto, horas_establecidas, valor_hora, plazo_estimado):
    try:
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        nombre_archivo = f"Proyecto{timestamp}.pdf"

        c = canvas.Canvas(
            nombre_archivo, pagesize=A4
        )  # Crear un nuevo PDF con el nombre generado
        width, height = A4  # Obtener dimensiones de la página

        # Agregar plantilla de fondo si existe
        ruta_template = "outputs/Template.png"  # Ajusta según tu carpeta
        if os.path.exists(ruta_template):  # Verificar si la plantilla existe
            template = ImageReader(ruta_template)  # Cargar la plantilla
            c.drawImage(
                template, 0, 0, width=width, height=height
            )  # Dibujar la plantilla como fondo

        # Agregar logo si existe
        ruta_logo = "outputs/logo.jpg"  # Ajusta según tu carpeta
        if os.path.exists(ruta_logo):  # Cargar el logo
            logo = ImageReader(ruta_logo)  # Cargar el logo
            c.drawImage(
                logo, width - 700, height - 750, preserveAspectRatio=True
            )  # Dibujar el logo en la esquina superior derecha

        # Agregar contenido al PDF
        c.setFont("Helvetica-Bold", 16)  # Título del reporte
        c.drawString(
            340, height - 395, f"{proyecto.upper()}"
        )  # Agregar título del proyecto

        # Agregar detalles del proyecto
        c.setFont("Helvetica-Bold", 14)
        c.drawString(
            400, height - 430, f"{horas_establecidas}"
        )  # Agregar horas establecidas
        c.drawString(400, height - 470, f"{valor_hora}")  # Agregar valor por hora
        c.drawString(400, height - 510, f"{plazo_estimado}")  # Agregar plazo estimado

        # Calcular y agregar el valor total del proyecto
        valor_total = int(horas_establecidas) * float(
            valor_hora
        )  # Calcular el valor total
        c.setFont("Helvetica-Bold", 20)  # Resaltar el valor total
        c.drawString(
            400, height - 550, f"{valor_total}"
        )  # Agregar el valor total al PDF
        c.save()
        print(f"✅ PDF '{nombre_archivo}' creado con éxito.")
        return nombre_archivo
    except Exception as e:
        print(f"❌ Error al generar: {e}")


generar_presupuesto(
    proyecto="Analista DataScience",
    horas_establecidas="416",
    valor_hora="35.00",
    plazo_estimado="2 Meses",
)
```

> ## 📐\*Guia de calibracion (𝑥:0 -𝕪:0)
>
> ### \* La ubicacion la tomamos desde la esquna superior Izquierda, donde (-𝕪) aumentarà en 👇🏼direccion decendente y a su vez (𝑥)aumentarà en 👉🏻direccion hacia la derecha.


```python
import os
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import letter
from reportlab.lib import colors


def generar_cuadrilla_superior():
    nombre_archivo = "CALIBRACION_SUPERIOR.pdf"
    c = canvas.Canvas(nombre_archivo, pagesize=letter)
    width, height = letter  # 612 x 792 (Ancho x Alto)

    # 1. Intentar cargar tu plantilla de fondo para referencia
    ruta_fondo = "outputs/Template.png"
    if os.path.exists(ruta_fondo):
        c.setFillAlpha(0.3)  # Ponemos el fondo clarito para ver la cuadrícula
        c.drawImage(ruta_fondo, 0, 0, width=width, height=height)
        c.setFillAlpha(1.0)

    # 2. Configuración de la Regla
    c.setFont("Helvetica", 7)

    # --- DIBUJAR LÍNEAS HORIZONTALES (Eje Y aumentando hacia abajo) ---
    # Recorremos desde 0 (arriba) hasta height (abajo)
    for i in range(0, int(height), 20):
        # Convertimos la coordenada de ReportLab para que el 0 sea arriba
        y_real = height - i

        if i % 100 == 0:
            c.setStrokeColor(colors.blue)  # Líneas principales cada 100px
            c.setLineWidth(1)
        else:
            c.setStrokeColor(colors.red)
            c.setLineWidth(0.2)

        c.line(0, y_real, width, y_real)
        # Escribimos el valor de Y (distancia desde el techo)
        c.setFillColor(colors.black)
        c.drawString(5, y_real + 2, f"Y={i}")
        c.drawString(width - 35, y_real + 2, f"Y={i}")

    # --- DIBUJAR LÍNEAS VERTICALES (Eje X aumentando a la derecha) ---
    for x in range(0, int(width), 20):
        if x % 100 == 0:
            c.setStrokeColor(colors.blue)
            c.setLineWidth(1)
        else:
            c.setStrokeColor(colors.red)
            c.setLineWidth(0.2)

        c.line(x, 0, x, height)
        c.setFillColor(colors.black)
        c.drawString(x + 2, 10, f"X={x}")  # Coordenada X abajo
        c.drawString(x + 2, height - 12, f"X={x}")  # Coordenada X arriba

    c.save()
    print(f"📏 Cuadrícula generada: '{nombre_archivo}'")
    print(
        "Nota: El 0 está arriba. Si un cuadro está en Y=400, está a 400 puntos del techo."
    )


generar_cuadrilla_superior()
```
